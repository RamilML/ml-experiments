```python
import sklearn as sk
import torch as pt
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from pandas.plotting import table
data = pd.read_csv('/content/raw.csv', sep = r'\s+', header = None)
test_data = pd.read_csv('/content/test.csv', sep = r'\s+', header = None)

featurs_cols = [f'enr_{i}' for i in range(1, 40)]
data.columns = ['k_inf', 'pppf'] + featurs_cols
test_data.columns = ['k_inf','pppf'] + featurs_cols
data = data.drop(columns='pppf')
print(data.head())
print(data.columns.to_list())
print(data.shape)

y_train = data[['k_inf']]
X_train = data.drop(columns = ['k_inf'])

y_test = test_data[['k_inf']]
X_test = test_data.drop(columns = ['k_inf'])

print(data.info())
print(test_data.info())
```

```python
from sklearn.model_selection import train_test_split
train_x, val_x, train_y, val_x = train_test_split(X_train, y_train, train_size= 90, test_size= 10)
```

```python
from sklearn.linear_model import LinearRegression
lin = LinearRegression().fit(train_x, train_y)
predict_lin = lin.predict(X_test)
Q1 = np.mean((y_test.iloc[:,0] - predict_lin[:,0])**2)
Q2 = np.mean((y_test.iloc[:,1] - predict_lin[:,1])**2)
Q = (Q1 + Q2)/2
print(f'Q1 = {Q1}')
print(f'Q2 = {Q2}')
print(f'Q = {Q}')
```
